```python
# Configuração da Chave de API
import os
from google import genai


# Utilize o Secrets do Google Colab para aplicar a API no código
from google.colab import userdata
Chave_API = userdata.get('GEMINI_API_KEY')


if not Chave_API:
    print("ERRO: A variável de ambiente GEMINI_API_KEY não está configurada.")
else:
    # Inicializa o cliente da API
    client = genai.Client(api_key=Chave_API)
    print("Cliente da API inicializado com sucesso!")
```

```python
# Função principal
def gerar_texto_com_llm(prompt_inicial: str, modelo: str = 'gemini-2.5-flash'):
    """
    Utiliza a API do Google GenAI para gerar texto com base em um prompt.

    Args:
        prompt_inicial: O texto de entrada para o modelo.
        modelo: O nome do modelo a ser utilizado (padrão: gemini-2.5-flash).

    Returns:
        O texto gerado pelo modelo como uma string.
    """

    # 1. Utilizar a API de LLM para gerar um texto
    try:
        response = client.models.generate_content(
            model=modelo,
            contents=prompt_inicial
        )

        # Retorna o texto gerado
        return response.text

    except Exception as e:
        return f"Ocorreu um erro ao gerar o conteúdo: {e}"

# 2. Desenvolver um script Python que leia um prompt e retorne o texto gerado
if __name__ == '__main__':
    # Exemplo de prompt
    prompt_usuario = "Escreva um parágrafo curto e criativo sobre uma história que se passa no universo do game CyberPunk2077"

    print("--- 🤖 Iniciando Geração de Texto ---")
    print(f"**Prompt Fornecido:**\n> {prompt_usuario}\n")

    texto_gerado = gerar_texto_com_llm(prompt_usuario)

    print("\n--- ✨ Texto Gerado pelo Modelo ---")
    print(texto_gerado)
    print("-----------------------------------")
```
