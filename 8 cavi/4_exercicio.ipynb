```python
# Configuração da Chave de API
import os
from google import genai


# Utilize o Secrets do Google Colab para aplicar a API no código
from google.colab import userdata
Chave_API = userdata.get('GEMINI_API_KEY')


if not Chave_API:
    print("ERRO: A variável de ambiente GEMINI_API_KEY não está configurada.")
else:
    # Inicializa o cliente da API
    client = genai.Client(api_key=Chave_API)
    print("Cliente da API inicializado com sucesso!")
```

```python
# Função principal
def resolver_problema_com_llm(pergunta_ou_problema: str, modelo: str = 'gemini-2.5-flash'):
    """
    Utiliza a API do Google GenAI para responder a uma pergunta ou resolver um problema.

    Args:
        pergunta_ou_problema: O prompt que contém a pergunta ou o problema a ser resolvido.
        modelo: O nome do modelo a ser utilizado (padrão: gemini-2.5-flash).

    Returns:
        A resposta ou solução fornecida pelo modelo.
    """

    # 1. Utilizar a API de LLM para responder a uma pergunta ou resolver um problema
    try:
        response = client.models.generate_content(
            model=modelo,
            contents=pergunta_ou_problema
        )

        # Retorna o texto gerado
        return response.text

    except Exception as e:
        return f"Ocorreu um erro ao gerar o conteúdo: {e}"

# 2. Desenvolver um script Python que leia uma pergunta e retorne a resposta
if __name__ == '__main__':

    # Exemplo de prompt
    problema_exemplo = "Estou com dificuldades para passar do boss Malenia no jogo Elden Ring, quais ações poderiam tornar esse desafio mais fácil?"

    print("--- ❓ Iniciando Resolução de Problemas ---")
    print(f"**Problema Fornecido:**\n> {problema_exemplo}\n")

    solucao_gerada = resolver_problema_com_llm(problema_exemplo)

    print("\n--- ✅ Solução/Resposta Gerada pelo Modelo ---")
    print(solucao_gerada)
    print("---------------------------------------------")
```
