```python
# Configuração e Inicialização do Cliente
import os
import google.generativeai as genai
from google.api_core.exceptions import GoogleAPICallError as APIError # Usando GoogleAPICallError para tratamento de erros da API

# Utilize o Secrets do Google Colab para aplicar a API no código
from google.colab import userdata
Chave_API = userdata.get('GEMINI_API_KEY')


if not Chave_API:
    print("ERRO: A variável de ambiente GEMINI_API_KEY não está configurada.")
else:
    # A chave da API será configurada explicitamente na função iniciar_agente_virtual
    print("Chave da API recuperada com sucesso.")
    client = None # Define client como None, pois não é usado para o GenerativeModel
```

```python
# Inicialização do Agente Virtual (Objeto Chat)

# 1. Utilizar a API de LLM para interagir com o usuário em uma conversa.
def iniciar_agente_virtual(client_obj): # 'client_obj' não é diretamente usado para criar o GenerativeModel
    """Inicializa um novo objeto de chat (Agente Virtual)."""

    # O modelo 'gemini-2.5-flash' é ideal para conversas rápidas
    # system_instruction define a personalidade do agente

    # Garante que a Chave_API esteja definida e configurada para genai.
    global Chave_API
    if not ('Chave_API' in globals() and Chave_API):
        raise ValueError("A chave da API ('Chave_API') não está definida ou acessível.")

    # Configura a chave da API globalmente para o módulo genai.
    genai.configure(api_key=Chave_API)

    # Usa genai.GenerativeModel para criar o modelo com system_instruction
    model = genai.GenerativeModel(
        model_name = "gemini-2.5-flash",
        system_instruction = "Você é o 'AstroBot', um assistente virtual amigável especializado em Jogos de videogame e PC. Mantenha as respostas focadas nesse tema e seja divertido."
    )
    # Inicia uma sessão de chat a partir do modelo com um histórico vazio
    chat = model.start_chat(history=[])
    return chat

# Inicializa o chat fora da função principal para manter o estado da conversa
try:
    # 'client' é esperado para ser definido a partir da célula anterior, mas pode ser None.
    # Passamos ele para manter a assinatura da função, embora não seja usado internamente agora.
    agente_virtual = iniciar_agente_virtual(client)
    print("\n✅ AstroBot Inicializado! (Agente Virtual criado com histórico)")
except NameError as e:
    # Isso irá capturar se 'client' não estiver definido.
    print(f"\n⚠️ Erro: Variável essencial não encontrada. Por favor, certifique-se de que a Chave_API está configurada e 'client' está definido na Célula 1 antes de prosseguir. Detalhes: {e}")
except APIError as e:
    print(f"\n⚠️ Erro de API ao inicializar o chat: {e}")
except Exception as e:
    print(f"\n⚠️ Ocorreu um erro inesperado durante a inicialização do AstroBot: {e}")
```

```python
# Loop de Conversa Interativa
def conversar_com_agente(agente_virtual):
    """
    Desenvolve um loop de conversação interativa.
    Lê a entrada do usuário e retorna a resposta do agente.
    """
    print("\n--- 💬 Início da Conversa com o AstroBot ---")
    print("AstroBot: Olá! Sou o AstroBot. Podemos falar sobre quaisquer jogos atuais e antigos. Digite 'sair' para encerrar.")

    while True:
        # 2. Desenvolver um script Python que leia entradas do usuário
        entrada_usuario = input("\nVocê: ")

        if entrada_usuario.lower() == 'sair':
            print("\nAstroBot: Até a próxima jornada espacial! 🚀")
            break

        if not entrada_usuario.strip():
            continue

        try:
            # Envia a mensagem do usuário e recebe a resposta, atualizando o histórico
            response = agente_virtual.send_message(entrada_usuario)

            # Retorna a resposta do agente virtual
            print(f"AstroBot: {response.text}")

        except Exception as e:
            print(f"Erro na comunicação com a API: {e}")
            break

# Inicia a conversa
if 'agente_virtual' in locals():
    conversar_com_agente(agente_virtual)

# Opcional: Mostra o histórico da conversa após o fim
if 'agente_virtual' in locals():
    print("\n--- 📜 Histórico Completo da Conversa ---")
    for message in agente_virtual.history: # Corrigido: Acessando o atributo 'history' diretamente
        # Exibe o papel (user ou model) e o texto da mensagem
        print(f"[{message.role.capitalize()}]: {message.parts[0].text}")
```
